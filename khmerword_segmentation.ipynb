# Khmer word segmentation

All segmentation and POS rules are implemented in `model.py`. This notebook only loads the trained model and calls the reusable functions.

```python
# Import the model loader and the main text-segmentation function.
from model import load_model, segment

# Load the Khmer dictionary and build the searchable trie.
model = load_model()

# Confirm how many dictionary entries are available.
print(f"Loaded {model['word_count']:,} dictionary words")
```

```python
# This is the Khmer sentence that will be segmented.
input_text = "ក្មេងម្នាក់នេះ មិនខ្លាចការនឿយហត់ទេ."

# Split the sentence into tokens and assign POS/status information.
tokens = segment(input_text, model)

# Keep the first occurrence of each word for a compact summary table.
unique_tokens = list({token["word"]: token for token in tokens}.values())

def display_width(text):
    """Calculate terminal display width, including Khmer characters."""
    # Khmer combining marks do not use an additional display column.
    import unicodedata
    return sum(
        0 if unicodedata.category(character).startswith("M") or unicodedata.category(character) == "Cf"
        else 2 if unicodedata.east_asian_width(character) in {"W", "F"}
        else 1
        for character in str(text)
    )

def pad_cell(value, width):
    """Pad a table value to a consistent visual width."""
    # Convert values such as token numbers into text before padding.
    value = str(value)
    # Add only the spaces needed to reach the requested width.
    return value + " " * max(0, width - display_width(value))

def format_token(number, token):
    """Format one token as a readable table row."""
    return (
        f"{pad_cell(number, 4)}  {pad_cell(token['word'], 24)}  "
        f"{pad_cell(token['pos'], 24)}  {pad_cell(token['status'], 20)}"
    )

# Create the column headings used by the output table.
header = f"{pad_cell('No.', 4)}  {pad_cell('Word', 24)}  {pad_cell('Part of speech', 24)}  {pad_cell('Status', 20)}"
# Print the heading and a separator line with matching display width.
print(header)
print("-" * display_width(header))
# Print one formatted row for each unique token.
for number, token in enumerate(unique_tokens, start=1):
    print(format_token(number, token))

# Print summary information after the table.
print(f"\nTotal words: {len(tokens)}")
print(f"Unique words: {len(unique_tokens)}")
```
